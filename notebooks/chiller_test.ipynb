```python
import time
from src.devices.chiller import JulaboChiller

# Initialize Chiller
# Ensure the port matches your configuration (e.g., COM8)
chiller = JulaboChiller(port="COM8", baudrate=9600)

# Connect
if chiller.connect():
    print("Connected successfully!")
else:
    print("Failed to connect.")
```

```python
# Read Status and Temperatures
if chiller.ser and chiller.ser.is_open:
    status = chiller.get_status()
    print(f"Status: {status}")
    
    actual_temp = chiller.get_internal_temperature()
    print(f"Internal Temperature: {actual_temp}")

    external_temp = chiller.get_external_temperature()
    print(f"External Temperature: {external_temp}")
    
    setpoint = chiller.get_setpoint_temperature()
    print(f"Setpoint Temperature: {setpoint}")
```

```python
chiller.send_command("in_pv_00")
msg = chiller.read_response()
print(msg)
```

```python
# Test Setting Temperature
target_temp = 15.0
chiller.set_setpoint_temperature(target_temp)
print(f"Set temperature to {target_temp}")
time.sleep(1)
print(f"New Setpoint: {chiller.get_setpoint_temperature()}")
```
