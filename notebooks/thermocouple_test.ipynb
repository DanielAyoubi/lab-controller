```python
from src.devices.thermocouple import Thermocouple

thermo = Thermocouple(vendor_id=0x2177, product_id=0x0004)

if thermo.connect():
    try:
        for i in range(5):
            temp = thermo.get_temperature()
            if temp is None:
                print("No reading returned.")
            else:
                print(f"Current temperature: {temp:.2f} °C")
            time.sleep(1)
    finally:
        thermo.disconnect()
else:
    print("Thermocouple failed to connect.")
```
